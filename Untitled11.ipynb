```python
import os, cv2, time, glob, numpy as np, matplotlib.pyplot as plt
from mtcnn.mtcnn import MTCNN
from deepface import DeepFace

# Visualização
def imshow_bgr(title, img_bgr, size=10):
    assert img_bgr is not None, "Imagem vazia"
    plt.figure(figsize=(size, size*img_bgr.shape[0]/img_bgr.shape[1]))
    plt.imshow(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB))
    plt.axis('off'); plt.title(title); plt.show()

def draw_box_label(img, box, label):
    x,y,w,h = box
    cv2.rectangle(img, (x,y), (x+w,y+h), (0,255,0), 2)
    cv2.putText(img, label, (x, max(0,y-6)), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)

def safe_crop(img, x,y,w,h, m=0.20):
    H,W = img.shape[:2]; dx,dy = int(w*m), int(h*m)
    x1,y1 = max(0,x-dx), max(0,y-dy)
    x2,y2 = min(W,x+w+dx), min(H,y+h+dy)
    return img[y1:y2, x1:x2], (x1,y1,x2-x1,y2-y1)

# Detector MTCNN (CPU)
detector = MTCNN()

def detect_faces_bgr(img_bgr, min_size=24, min_conf=0.70):
    rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    dets = detector.detect_faces(rgb)
    out=[]
    for d in dets:
        x,y,w,h = d['box']; conf=d['confidence']
        if w>=min_size and h>=min_size and conf>=min_conf:
            out.append({'box':(x,y,w,h),'confidence':conf})
    return out

# Embeddings DeepFace (Facenet512) com compatibilidade
try:
    EMB_MODEL = DeepFace.build_model("Facenet512")
except Exception:
    EMB_MODEL = None

def face_embedding(face_bgr):
    face_rgb = cv2.cvtColor(face_bgr, cv2.COLOR_BGR2RGB)
    try:
        reps = DeepFace.represent(img_path=face_rgb, model_name="Facenet512",
                                  detector_backend="skip", enforce_detection=False,
                                  model=EMB_MODEL)
    except TypeError:
        reps = DeepFace.represent(img_path=face_rgb, model_name="Facenet512",
                                  detector_backend="skip", enforce_detection=False)
    emb = np.array(reps[0]["embedding"], dtype=np.float32)
    return emb

# Cosseno em NumPy (sem sklearn)
def cosine_similarity_np(q, G):
    qn = q / (np.linalg.norm(q, axis=1, keepdims=True) + 1e-12)
    Gn = G / (np.linalg.norm(G, axis=1, keepdims=True) + 1e-12)
    return (qn @ Gn.T)[0]
```

```python
def build_gallery_from_image(img_path, names_left2right, min_size=24, min_conf=0.70):

    img = cv2.imread(img_path); assert img is not None, f"Não achei: {img_path}"
    dets = detect_faces_bgr(img, min_size=min_size, min_conf=min_conf)
    assert len(dets) >= len(names_left2right), \
        f"Detectei {len(dets)} face(s), mas recebi {len(names_left2right)} nome(s). Afrouxe min_conf/min_size."
    dets = sorted(dets, key=lambda d: d['box'][0])  # esquerda→direita

    gallery = {}  # nome -> vetor 512D
    for i, name in enumerate(names_left2right):
        x,y,w,h = dets[i]['box']
        crop,_ = safe_crop(img, x,y,w,h)
        emb = face_embedding(crop)
        gallery[name] = emb.astype(np.float32)
    return gallery

def save_gallery_npz(path_npz, gallery_dict):
    names = np.array(list(gallery_dict.keys()))
    embs  = np.vstack([gallery_dict[n] for n in names])
    np.savez(path_npz, names=names, embs=embs)
    return path_npz

def load_gallery_npz(path_npz):
    data = np.load(path_npz, allow_pickle=False)
    names = data["names"].astype(str).tolist()
    embs  = data["embs"].astype(np.float32)
    return {n: e for n,e in zip(names, embs)}
```

```python
def recognize_with_gallery_memory(img_path, gallery, threshold=0.45, min_size=24, min_conf=0.70):
    img = cv2.imread(img_path); assert img is not None, f"Não achei: {img_path}"
    dets = detect_faces_bgr(img, min_size=min_size, min_conf=min_conf)
    names = list(gallery.keys())
    if not names:
        raise RuntimeError("Galeria vazia.")
    G = np.vstack([gallery[n] for n in names])  # (N,512)

    vis = img.copy()
    results = []
    for d in dets:
        x,y,w,h = d['box']
        crop, box_safe = safe_crop(img, x,y,w,h)
        q = face_embedding(crop).reshape(1,-1)
        sims = cosine_similarity_np(q, G)
        j = int(np.argmax(sims)); sim = float(sims[j])
        name = names[j] if sim >= threshold else "desconhecido"
        results.append((box_safe, name, sim))
        draw_box_label(vis, box_safe, f"{name} ({sim:.2f})")
    return vis, results
```

```python
cands = [p for ext in ("jpg","jpeg","png")
         for p in glob.glob(f"/content/**/*.{ext}", recursive=True)
         if "/content/galeria/" not in p]
for p in sorted(cands): print(" -", p)

img_path_multi = "/content/galeria/pessoas_A/IMG_0109.jpeg"   # caminho da imagem

# Nomes na ordem ESQUERDA → DIREITA
names4 = ["Graham O'Brien", "Doutor", "Ryan Sinclair", "Yasmin Khan"]

# Construir a galeria em memória (NÃO cria arquivos de recorte)
gallery_mem = build_gallery_from_image(img_path_multi, names4, min_size=20, min_conf=0.60)
print("Galeria (memória):", list(gallery_mem.keys()))

# persista a galeria como .npz
_ = save_gallery_npz("/content/galeria_mem.npz", gallery_mem)
```
